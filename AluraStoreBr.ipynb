### Importação dos dados



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_1.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_4.csv"

loja = pd.read_csv(url)
loja2 = pd.read_csv(url2)
loja3 = pd.read_csv(url3)
loja4 = pd.read_csv(url4)

loja.head()
```

#1. Análise do faturamento


```python
import matplotlib.pyplot as plt

# Calcular faturamento por loja
faturamento_loja1 = loja['Preço'].sum()
faturamento_loja2 = loja2['Preço'].sum()
faturamento_loja3 = loja3['Preço'].sum()
faturamento_loja4 = loja4['Preço'].sum()

# Dicionário com os resultados
faturamento_total = {
    'Loja 1': faturamento_loja1,
    'Loja 2': faturamento_loja2,
    'Loja 3': faturamento_loja3,
    'Loja 4': faturamento_loja4
}

# Exibir valores em formato monetário
print("Faturamento por Loja:")
for loja_nome, valor in faturamento_total.items():
    print(f'{loja_nome}: R$ {valor:,.2f}')

# Criar gráfico de barras
plt.figure(figsize=(10, 6))
bars = plt.bar(
    faturamento_total.keys(),
    faturamento_total.values(),
    color=['skyblue', 'lightgreen', 'salmon', 'gold']
)

# Adicionar rótulos e título
plt.title('Faturamento por Loja', fontsize=16, pad=20)
plt.xlabel('Lojas', fontsize=12)
plt.ylabel('Faturamento (R$)', fontsize=12)

# Adicionar valores em cima das barras
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height,
        f'R$ {height:,.2f}',
        ha='center',
        va='bottom',
        fontsize=10
    )

# Ajustar layout e mostrar gráfico
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

# 3. Média de Avaliação das Lojas

```python
import matplotlib.pyplot as plt
import numpy as np

# Função para calcular a média das avaliações
def calcular_media_avaliacoes(df, nome_loja):
    try:
        media = df['Avaliação da compra'].mean()
        return round(media, 2)  # Arredonda para 2 casas decimais
    except KeyError:
        print(f"Erro: Coluna 'Avaliação da compra' não encontrada em {nome_loja}.")
        return None

# Calcular médias para cada loja
medias = {
    'Loja 1': calcular_media_avaliacoes(loja, "Loja 1"),
    'Loja 2': calcular_media_avaliacoes(loja2, "Loja 2"),
    'Loja 3': calcular_media_avaliacoes(loja3, "Loja 3"),
    'Loja 4': calcular_media_avaliacoes(loja4, "Loja 4")
}

# Exibir resultados
print("\nMédia de Avaliações por Loja (1-5):")
for loja_nome, media in medias.items():
    if media is not None:
        print(f"{loja_nome}: {media}/5")

# Gráfico de barras
if all(media is not None for media in medias.values()):
    plt.figure(figsize=(10, 5))
    bars = plt.bar(
        medias.keys(),
        medias.values(),
        color=['#FF9AA2', '#FFB7B2', '#FFDAC1', '#E2F0CB'],  # Cores pastel
        edgecolor='gray'
    )

    # Configurações do gráfico
    plt.title('Satisfação do Cliente por Loja', fontsize=14, pad=15)
    plt.xlabel('Lojas', fontsize=12)
    plt.ylabel('Média das Avaliações (1-5)', fontsize=12)
    plt.ylim(0, 5.5)  # Limite do eixo Y

    # Adicionar valores nas barras
    for bar in bars:
        height = bar.get_height()
        plt.text(
            bar.get_x() + bar.get_width() / 2,
            height,
            f'{height}',
            ha='center',
            va='bottom',
            fontsize=11,
            weight='bold'
        )

    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()
else:
    print("Não foi possível gerar o gráfico devido a dados faltantes.")
```

# 4. Produtos Mais e Menos Vendidos

```python
import matplotlib.pyplot as plt

def analisar_produtos_por_loja(df, nome_loja, top_n=5):
    # Contar ocorrências de cada produto
    contagem_produtos = df['Produto'].value_counts()

    # Produtos mais vendidos (top N)
    top_produtos = contagem_produtos.head(top_n)

    # Produtos menos vendidos (bottom N)
    bottom_produtos = contagem_produtos.tail(top_n)

    # Exibir resultados
    print(f"\nLoja: {nome_loja}")
    print(f"\nTOP {top_n} Produtos Mais Vendidos:")
    print(top_produtos.to_frame('Quantidade Vendida'))

    print(f"\nTOP {top_n} Produtos Menos Vendidos:")
    print(bottom_produtos.to_frame('Quantidade Vendida'))

    # Plotar gráficos
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 5))
    fig.suptitle(f'Análise de Produtos - {nome_loja}', fontsize=14)

    # Gráfico dos mais vendidos
    top_produtos.plot(kind='bar', ax=ax1, color='green', alpha=0.7)
    ax1.set_title(f'Top {top_n} Mais Vendidos')
    ax1.set_ylabel('Quantidade Vendida')
    ax1.tick_params(axis='x', rotation=45)

    # Gráfico dos menos vendidos
    bottom_produtos.plot(kind='bar', ax=ax2, color='red', alpha=0.7)
    ax2.set_title(f'Top {top_n} Menos Vendidos')
    ax2.set_ylabel('Quantidade Vendida')
    ax2.tick_params(axis='x', rotation=45)

    plt.tight_layout()
    plt.show()

# Exemplo de uso (substitua com seus DataFrames reais)
analisar_produtos_por_loja(loja, "Loja 1")
analisar_produtos_por_loja(loja2, "Loja 2")
analisar_produtos_por_loja(loja3, "Loja 3")
analisar_produtos_por_loja(loja4, "Loja 4")
```

# 5. Frete Médio por Loja


```python
import matplotlib.pyplot as plt
import numpy as np

# Função para calcular o frete médio com tratamento de erros
def calcular_frete_medio(df, nome_loja):
    try:
        if 'Frete' not in df.columns:
            raise KeyError(f"Coluna 'Frete' não encontrada em {nome_loja}")

        media = df['Frete'].mean()
        return round(media, 2)  # Arredonda para 2 casas decimais

    except Exception as e:
        print(f"Erro em {nome_loja}: {str(e)}")
        return None

# Dicionário para armazenar os resultados
frete_medio = {
    'Loja 1': calcular_frete_medio(loja, "Loja 1"),
    'Loja 2': calcular_frete_medio(loja2, "Loja 2"),
    'Loja 3': calcular_frete_medio(loja3, "Loja 3"),
    'Loja 4': calcular_frete_medio(loja4, "Loja 4")
}

# Exibir resultados no terminal
print("\nCusto Médio de Frete por Loja (R$):")
for loja_nome, valor in frete_medio.items():
    if valor is not None:
        print(f"{loja_nome}: R$ {valor:.2f}")

# Gráfico de barras (apenas para lojas com dados válidos)
lojas_validas = [k for k, v in frete_medio.items() if v is not None]
valores_validos = [v for v in frete_medio.values() if v is not None]

if lojas_validas:
    plt.figure(figsize=(10, 6))
    bars = plt.bar(
        lojas_validas,
        valores_validos,
        color=plt.cm.Pastel1(range(len(lojas_validas))))  # Cores pastel


    # Configurações do gráfico
    plt.title('Custo Médio de Frete por Loja', fontsize=14, pad=15)
    plt.xlabel('Lojas', fontsize=12)
    plt.ylabel('Valor do Frete (R$)', fontsize=12)

    # Adicionar valores em cima das barras
    for bar in bars:
        height = bar.get_height()
        plt.text(
            bar.get_x() + bar.get_width()/2,
            height,
            f'R$ {height:.2f}',
            ha='center',
            va='bottom',
            fontsize=11,
            weight='bold'
        )

    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()
else:
    print("Nenhum dado válido para gerar o gráfico.")
```

```python
import matplotlib.pyplot as plt

frete_medio = {
    'Loja 1': calcular_frete_medio(loja, "Loja 1"),
    'Loja 2': calcular_frete_medio(loja2, "Loja 2"),
    'Loja 3': calcular_frete_medio(loja3, "Loja 3"),
    'Loja 4': calcular_frete_medio(loja4, "Loja 4")
}

# Filtrar lojas com dados válidos
lojas_validas = [k for k, v in frete_medio.items() if v is not None]
valores_validos = [v for v in frete_medio.values() if v is not None]

if not lojas_validas:
    print("Nenhum dado válido para gerar o gráfico.")
else:
    # Configurações do gráfico
    cores = ['#FF6B6B', '#4ECDC4', '#45B7D1', '#FFA07A']
    explode = [0.05] * len(lojas_validas)  # Pequeno destaque para todas as fatias

    plt.figure(figsize=(10, 8))

    # Criar gráfico de pizza com valores absolutos e percentuais
    patches, texts, autotexts = plt.pie(
        valores_validos,
        labels=[f'{loja}\nR$ {valor:.2f}' for loja, valor in zip(lojas_validas, valores_validos)],
        autopct='%1.1f%%',
        startangle=90,
        colors=cores,
        explode=explode,
        shadow=True,
        wedgeprops={'edgecolor': 'white', 'linewidth': 1},
        textprops={'fontsize': 10, 'weight': 'bold'}
    )

    # Melhorar a legibilidade dos percentuais
    for autotext in autotexts:
        autotext.set_color('black')
        autotext.set_fontsize(10)

    # Título e legenda
    plt.title('Distribuição do Custo Médio de Frete por Loja\n', fontsize=14, pad=20)
    plt.legend(
        patches,
        lojas_validas,
        title="Lojas",
        loc="center left",
        bbox_to_anchor=(1, 0.5))

    plt.tight_layout()
    plt.show()
```
